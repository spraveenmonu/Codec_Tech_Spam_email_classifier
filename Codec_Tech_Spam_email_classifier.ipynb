### Spam Email Classifier

This notebook will demonstrate building a spam email classifier using TF-IDF and Naive Bayes, covering text cleaning, vectorization, training, and evaluation.

```python
import pandas as pd

# For demonstration purposes, let's assume the dataset is available as 'spam.csv'.
# You might need to upload it to your Colab environment or adjust the path.
# A common dataset is the 'SMS Spam Collection Dataset' available on Kaggle or UCI.
try:
    df = pd.read_csv('spam.csv', encoding='latin-1')
    # The dataset often has unnamed columns; we'll drop them and rename relevant ones.
    df = df.iloc[:, :2]
    df.columns = ['label', 'message']

    print("Dataset loaded successfully:")
    display(df.head())
    print("\nDataset information:")
    display(df.info())

except FileNotFoundError:
    print("Error: 'spam.csv' not found. Please upload the dataset or provide the correct path.")
    print("You can download the 'SMS Spam Collection Dataset' from Kaggle or UCI.")
```

### Data Preprocessing

Before training the model, we need to clean the text data and convert it into a numerical format that the machine learning model can understand. This typically involves:
1.  **Text Cleaning**: Lowercasing, removing punctuation, stop words, and potentially stemming/lemmatization.
2.  **Vectorization**: Converting text into numerical feature vectors, often using TF-IDF (Term Frequency-Inverse Document Frequency).

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download necessary NLTK data (if not already downloaded)
try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')

# Initialize stemmer and stopwords
stemmer = PorterStemmer()
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower() # Lowercase
    text = re.sub(r'[^a-z0-9\s]', '', text) # Remove punctuation
    words = text.split() # Tokenize
    words = [stemmer.stem(word) for word in words if word not in stop_words] # Remove stop words and stem
    return ' '.join(words)

# Apply text cleaning
if 'df' in locals() and not df.empty:
    df['cleaned_message'] = df['message'].apply(clean_text)
    print("\nMessages after cleaning:")
    display(df[['message', 'cleaned_message']].head())
else:
    print("DataFrame 'df' is not available for cleaning. Please ensure the dataset is loaded.")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=5000) # Limit features to prevent overfitting

# Apply TF-IDF Vectorization
if 'df' in locals() and not df.empty and 'cleaned_message' in df.columns:
    X = tfidf_vectorizer.fit_transform(df['cleaned_message'])
    y = df['label'].apply(lambda x: 1 if x == 'spam' else 0) # Convert labels to numerical (spam=1, ham=0)

    print("\nShape of TF-IDF feature matrix:", X.shape)
    print("\nFirst 5 TF-IDF features for the first message (sparse format):")
    display(X[0, :5])
    print("\nLabels (numerical):")
    display(y.head())
else:
    print("DataFrame 'df' or 'cleaned_message' column is not available for TF-IDF vectorization.")
```
